```python
import json
import os

print(os.getcwd())

# ── Paths ──────────────────────────────────────────────────────────────────────
SPLITS_DIR        = '//dartfs/rc/lab/B/BhattacharyaI/Results/nnUNet_data/nnUNet_processed/Dataset999_AutoPet/splits_final.json'

# Source directories
HARD_LABELS_DIR   = '//dartfs/rc/lab/B/BhattacharyaI/Public_Datasets/Autopet_III_nnunet_raw/Dataset888_AutoPet/labelsTr'
PSEUDO_LABELS_DIR = '//dartfs/rc/lab/B/BhattacharyaI/Results/nnUNet_data/nnUNet_results/Dataset999_AutoPet/autoPET3_Trainer__nnUNetResEncUNetLPlansMultiTalent__3d_fullres/fold_1/train_predictions'
VAL_LABELS_DIR    = '//dartfs/rc/lab/B/BhattacharyaI/Results/nnUNet_data/nnUNet_results/Dataset999_AutoPet/autoPET3_Trainer__nnUNetResEncUNetLPlansMultiTalent__3d_fullres/val_labels'

# Destination
DATASET_ID        = 901   # <-- CHANGE AS NEEDED
NNUNET_RAW_DIR    = '//dartfs/rc/lab/B/BhattacharyaI/Results/nnUNet_data/nnUNet_raw'
NEW_DATASET_DIR   = os.path.join(NNUNET_RAW_DIR, f'Dataset{DATASET_ID:03d}_AutoPet')
OUTPUT_LABELS_DIR = os.path.join(NEW_DATASET_DIR, 'labelsTr')

# ── Load splits ────────────────────────────────────────────────────────────────
with open(SPLITS_DIR, 'r') as f:
    splits = json.load(f)

labelled   = set(splits[0]['train'])
all_files  = set(splits[6]['train'])
unlabelled = all_files - labelled
val_files  = set(splits[0]['val'])

# ── Create output directory ────────────────────────────────────────────────────
os.makedirs(OUTPUT_LABELS_DIR, exist_ok=True)
print(f'Created dataset directory: {NEW_DATASET_DIR}')

# ── Helper ─────────────────────────────────────────────────────────────────────
def make_symlink(src: str, dst: str) -> None:
    """
    Create a symlink at `dst` pointing to `src`.
    Skips gracefully if the source does not exist or the link already exists.
    """
    if not os.path.exists(src):
        print(f'  [MISSING]  {os.path.basename(src)}')
        return
    if os.path.exists(dst) or os.path.islink(dst):
        print(f'  [EXISTS]   {os.path.basename(dst)}')
        return
    os.symlink(src, dst)


def link_label(fname: str, source_dir: str) -> None:
    """Symlink a single label file from `source_dir` into OUTPUT_LABELS_DIR."""
    filename = fname + '.nii.gz'
    src = os.path.join(source_dir, filename)
    dst = os.path.join(OUTPUT_LABELS_DIR, filename)
    make_symlink(src, dst)


# ── 1. Hard (ground-truth) labels  ────────────────────────────────────────────
print('\n── Linking hard labels ──')
for fname in labelled:
    link_label(fname, HARD_LABELS_DIR)
print(f'  Processed {len(labelled)} hard label(s).')

# ── 2. Pseudo labels (unlabelled cases) ───────────────────────────────────────
print('\n── Linking pseudo labels ──')
for fname in unlabelled:
    link_label(fname, PSEUDO_LABELS_DIR)
print(f'  Processed {len(unlabelled)} pseudo label(s).')

# ── 3. Validation labels ──────────────────────────────────────────────────────
print('\n── Linking validation labels ──')
for fname in val_files:
    link_label(fname, VAL_LABELS_DIR)
print(f'  Processed {len(val_files)} validation label(s).')

# ── Summary ───────────────────────────────────────────────────────────────────
total = len(labelled) + len(unlabelled) + len(val_files)
print(f'\nDone! Created up to {total} symlink(s) in:\n  {OUTPUT_LABELS_DIR}')
```

```python
import pandas as pd
import numpy as np

# Merge the two dataframes on filename
df1, df2 = summary_dfs[0], summary_dfs[1]
# adjust the filename in df1 
df1['filename'] = df1['filename'].str.replace('_total_uncertainty', '', regex=False)


merged = df1.merge(df2, on='filename', suffixes=('_m1', '_m2'))

uncertainty_cols = [
    'average_uncertainty',
    'average_uncertainty_above_0',
    'average_uncertainty_predicted',
    'average_uncertainty_non_predicted'
]

# Calculate absolute differences for each uncertainty metric
for col in uncertainty_cols:
    merged[f'diff_{col}'] = abs(merged[f'{col}_m1'] - merged[f'{col}_m2'])

# Overall uncertainty similarity score (mean of all diffs)
diff_cols = [f'diff_{col}' for col in uncertainty_cols]
merged['mean_uncertainty_diff'] = merged[diff_cols].mean(axis=1)

# Average dice across both models
merged['mean_dice'] = (merged['dice_m1'] + merged['dice_m2']) / 2
merged['dice_diff'] = abs(merged['dice_m1'] - merged['dice_m2'])

# --- Interesting cases ---

print("=== Both high dice, similar uncertainty ===")
high_dice_similar = merged[merged['mean_dice'] > 0.8].nsmallest(5, 'mean_uncertainty_diff')
print(high_dice_similar[['filename', 'mean_dice', 'mean_uncertainty_diff'] + diff_cols])

print("\n=== Both high dice, dissimilar uncertainty ===")
high_dice_dissimilar = merged[merged['mean_dice'] > 0.8].nlargest(5, 'mean_uncertainty_diff')
print(high_dice_dissimilar[['filename', 'mean_dice', 'mean_uncertainty_diff'] + diff_cols])

print("\n=== One model high dice, other low dice (interesting disagreements) ===")
merged['dice_diff_signed'] = merged['dice_m1'] - merged['dice_m2']
disagreements = merged.reindex(merged['dice_diff'].nlargest(5).index)
print(disagreements[['filename', 'dice_m1', 'dice_m2', 'dice_diff', 'mean_uncertainty_diff']])

pass
```

```python
import shutil
import nibabel as nib
import os
import json

def move_case_to_uncertainty_examples(case_name, save_subdir=None, TTA_only=True):
    """
    Function to copy a given case for example viewing. Copies the uncertainty map,
    the image itself, the predicted labels, and the true labels for both ENS and TTA.
    Args:
        case_name (str): The case filename e.g. 'PETCT_001.nii.gz'
        save_subdir (str): Optional subdirectory name under SAVE_DIR. Defaults to case_name (no extension).
    """
    # --- Set up save directory ---
    if save_subdir is None:
        save_subdir = case_name.replace('.nii.gz', '').replace('.nii', '')
    
    case_save_dir = os.path.join(SAVE_DIR, save_subdir)
    labels = ['ENS', 'TTA']

    # --- Define all expected output files ---
    expected_files = (
        [f'uncertainty_map_{lab}_{case_name}' for lab in labels] +
        [f'predicted_{lab}_{case_name}'       for lab in labels] +
        [f'true_labels_{case_name}']                             +
        [f'image_{case_name.replace(".nii.gz", "_0001.nii.gz")}'] +
        ['metrics.json']
    )

    # --- Check if already done ---
    if os.path.exists(case_save_dir):
        missing = [f for f in expected_files if not os.path.exists(os.path.join(case_save_dir, f))]
        if not missing:
            print(f"⏭ Skipping {case_name} — all files already present in {case_save_dir}")
            return case_save_dir
        else:
            print(f"⚠ Directory exists but missing files: {missing} — continuing copy...")
    
    os.makedirs(case_save_dir, exist_ok=True)

    if TTA_only:
        i = 1
        
        print(f"Processing case: {case_name} (TTA only)")
        case_name_unc = case_name.replace('.nii.gz', '_uncertainty_uncertainty.nii.gz')
        unc_dir = UNCERTAINTY_MAPS_DIRS[i]
        src = os.path.join(unc_dir, case_name_unc)
        dst = os.path.join(case_save_dir, f'uncertainty_map_{labels[i]}_{case_name}')
        if os.path.exists(src):
            shutil.copy2(src, dst)
            print(f"✓ Copied uncertainty map [{labels[i]}]: {src}")
        else:
            try:
                src_alt = src.replace('.nii.gz', '_total_uncertainty.nii.gz')
                if os.path.exists(src_alt):
                    shutil.copy2(src_alt, dst)
                    print(f"✓ Copied uncertainty map [{labels[i]}]: {src_alt}")
                else:
                    print(f"✗ Uncertainty map not found [{labels[i]}]: {src} or {src_alt}")
            except Exception as e:
                print(f"Error copying uncertainty map [{labels[i]}]: {e}")
            print(f"✗ Uncertainty map not found [{labels[i]}]: {src}")

        pred_dir = PREDICTED_LABELS_DIRS[i]
        src = os.path.join(pred_dir, case_name)
        dst = os.path.join(case_save_dir, f'predicted_{labels[i]}_{case_name}')
        if os.path.exists(src):
            shutil.copy2(src, dst)
            print(f"✓ Copied predicted labels [{labels[i]}]: {src}")
        else:
            print(f"✗ Predicted labels not found [{labels[i]}]: {src}")

    else:
        print(f"Processing case: {case_name} (ENS and TTA)")
    # --- Copy uncertainty maps ---
        for i, unc_dir in enumerate(UNCERTAINTY_MAPS_DIRS):
            if i == 0:
                case_name_unc = case_name.replace('.nii.gz', '_total_uncertainty.nii.gz')
            else:
                case_name_unc = case_name.replace('.nii.gz', '_uncertainty_uncertainty.nii.gz')
            src = os.path.join(unc_dir, case_name_unc)
            dst = os.path.join(case_save_dir, f'uncertainty_map_{labels[i]}_{case_name}')
            if os.path.exists(src):
                shutil.copy2(src, dst)
                print(f"✓ Copied uncertainty map [{labels[i]}]: {src}")
            else:
                try:
                    src_alt = src.replace('.nii.gz', '_total_uncertainty.nii.gz')
                    if os.path.exists(src_alt):
                        shutil.copy2(src_alt, dst)
                        print(f"✓ Copied uncertainty map [{labels[i]}]: {src_alt}")
                    else:
                        print(f"✗ Uncertainty map not found [{labels[i]}]: {src} or {src_alt}")
                except Exception as e:
                    print(f"Error copying uncertainty map [{labels[i]}]: {e}")
                print(f"✗ Uncertainty map not found [{labels[i]}]: {src}")

        # --- Copy predicted labels ---
        for i, pred_dir in enumerate(PREDICTED_LABELS_DIRS):
            src = os.path.join(pred_dir, case_name)
            dst = os.path.join(case_save_dir, f'predicted_{labels[i]}_{case_name}')
            if os.path.exists(src):
                shutil.copy2(src, dst)
                print(f"✓ Copied predicted labels [{labels[i]}]: {src}")
            else:
                print(f"✗ Predicted labels not found [{labels[i]}]: {src}")

    # --- Copy true labels ---
    src = os.path.join(TRUE_LABELS_DIR.replace('imagesTr', 'labelsTr'), case_name)
    dst = os.path.join(case_save_dir, f'true_labels_{case_name}')
    if os.path.exists(src):
        shutil.copy2(src, dst)
        print(f"✓ Copied true labels: {src}")
    else:
        print(f"✗ True labels not found: {src}")

    # --- Copy raw image ---
    image_name = case_name.replace('.nii.gz', '_0001.nii.gz')
    images_dir = TRUE_LABELS_DIR.replace('labelsTr', 'imagesTr')
    src = os.path.join(images_dir, image_name)
    dst = os.path.join(case_save_dir, f'image_{image_name}')
    if os.path.exists(src):
        shutil.copy2(src, dst)
        print(f"✓ Copied raw image: {src}")
    else:
        print(f"✗ Raw image not found: {src}")

    # --- Write metrics.json ---
    metrics = {'case': case_name}
    for i, (lab, summary_df) in enumerate(zip(labels, summary_dfs)):
        row = summary_df[summary_df['filename'] == case_name]
        if not row.empty:
            row = row.iloc[0]
            metrics[lab] = {
                'dice'                              : float(row['dice']),
                'average_uncertainty'               : float(row['average_uncertainty']),
                'average_uncertainty_above_0'       : float(row['average_uncertainty_above_0']),
                'average_uncertainty_predicted'     : float(row['average_uncertainty_predicted']),
                'average_uncertainty_non_predicted' : float(row['average_uncertainty_non_predicted']),
            }
        else:
            metrics[lab] = f'No matching row found in summary_df for {case_name}'
            print(f"⚠ No metrics found for {case_name} in {lab} summary_df")

    # Add diffs if both found
    if isinstance(metrics.get('ENS'), dict) and isinstance(metrics.get('TTA'), dict):
        metrics['diff'] = {
            key: round(abs(metrics['ENS'][key] - metrics['TTA'][key]), 6)
            for key in metrics['ENS']
            if key != 'dice'  # dice diff kept separate for clarity
        }
        metrics['diff']['dice'] = round(abs(metrics['ENS']['dice'] - metrics['TTA']['dice']), 6)

    metrics_path = os.path.join(case_save_dir, 'metrics.json')
    with open(metrics_path, 'w') as f:
        json.dump(metrics, f, indent=4)
    print(f"✓ Wrote metrics to: {metrics_path}")

    print(f"\nAll files saved to: {case_save_dir}")
    return case_save_dir
```
